## **Exercise 05 : Boxplot**

```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

df = pd.read_csv('../data/ab-test.csv')

before = df.loc[(df['time'] == 'before')]
after = df.loc[(df['time'] == 'after')]

fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(15, 8))

after.boxplot(ax = ax1, column='diff', by=['group'], medianprops={"linewidth": 2, "color": "r"}, \
              boxprops={"linewidth": 3, "color": "g"}, showmeans=True, whiskerprops={"color": "r"},\
              capprops={"color": "r"})
ax1.set_title("after",fontsize=12)
ax1.set_xticklabels(['control', 'test'])
labels = ax1.get_yticks()

before.boxplot(ax = ax2, column='diff', by=['group'], medianprops={"linewidth": 2, "color": "r"}, \
               boxprops={"linewidth": 3, "color": "g"}, showmeans=True, whiskerprops={"color": "r"},\
               capprops={"color": "r"})
ax2.set_title("before",fontsize=12)
ax2.set_yticks(labels)
ax2.set_yticklabels([])
ax2.set_xticklabels(['control', 'test'])

fig.suptitle('Lag before labs deadlines. After and before the Newsfeed', fontsize=15)
plt.show()
```
