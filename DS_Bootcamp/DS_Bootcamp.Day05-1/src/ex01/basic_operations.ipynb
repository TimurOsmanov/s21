### **01.1 create a dataframe views with two columns: datetime and user by reading feed-views.log**

```python
import pandas as pd
df = pd.read_csv("../../datasets/feed-views-semicolon.log", index_col=None, delimiter=';')
# convert the datetime to the datetime64[ns] Dtype
df.date_time = pd.to_datetime(df.date_time)
# extract the year, month, day, hour, minute, and second from the values of that column to the new columns
df['year'] = df.date_time.dt.year
df['month'] = df.date_time.dt.month
df['day'] = df.date_time.dt.day
df['hour'] = df.date_time.dt.hour
df['minute'] = df.date_time.dt.minute
df['second'] = df.date_time.dt.second
df
```

### **01.2 create the new column daytime**

```python
# you need to assign the particular time of day value if an hour is within a particular interval, 
# for example, afternoon if the hour is larger than 11 and less or equal to 17
# 0 – 3.59 night, 4 – 6.59 early morning, 7 – 10.59 morning, 11 – 16.59 afternoon, 17 – 19.59 early evening, 20 – 23.59 evening
# use the method cut to solve this subtask
bins = [-1, 3, 6, 10, 16, 19, 24]
# means (-1,3], (3,6], (3,4] ...
labels = ['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening']
df['daytime'] = pd.cut(df.hour, bins=bins, labels=labels)
# assign the column user as the index
df.set_index('user', inplace=True)
df
```
