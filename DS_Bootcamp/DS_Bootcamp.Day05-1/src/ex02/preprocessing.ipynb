### **02.1 download and read the CSV file and make ID the index column**

```python
import pandas as pd
df = pd.read_csv("../../datasets/auto.csv", index_col=0)
df
```

### **02.2 count the number of observations using the method count()**

```python
df.count()
```

### **02.3 drop the duplicates, taking into account only the following columns: CarNumber, Make_n_model, Fines**

```python
# between the two equal observations, you need to choose the last
df.drop_duplicates(['CarNumber','Make_n_model','Fines'], keep='last', inplace=True)
df
```

```python
# check the number of observations again
df.count()
```

### **02.4 work with missing values**

```python
# check how many values are missing from each column
df.isnull().sum()
```

```python
# drop all the columns with over 500 missing values using the argument thresh, check how many missing values are in each column
df.dropna(thresh=500 ,axis=1, inplace=True)
df.isnull().sum()
```

```python
# replace all the missing values in the Refund column with the previous value in that column for that cell, use the argument method, 
# check how many values are missing from each column
df['Refund'] = df['Refund'].ffill()
df.isnull().sum()
```

```python
# replace all the missing values in the Fines column with the mean value of this column 
# (exclude NA/null values when computing the mean value), check how many values are missing from each column
df['Fines'] = df['Fines'].fillna(df['Fines'].mean())
df.isnull().sum()
```

### **02.5 split and parse the make and model**

```python
# use the method apply both for splitting and for extracting the values to the new columns Make and Model
df['Make_n_model'] = df['Make_n_model'].apply(lambda x: x.split())                          
df
```

```python
df[['Make', 'Model']] = df['Make_n_model'].apply(lambda x: pd.Series([i for i in x]))
df
```

```python
# drop the column Make_n_model
del df['Make_n_model']
df
```

```python
# save the dataframe in the JSON file auto.json in the format below:
```

```python
df.to_json('../../datasets/auto.json', orient="records", double_precision=1)
```
